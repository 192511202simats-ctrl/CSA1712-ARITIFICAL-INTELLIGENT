```python
from collections import deque

# Goal State
goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]

# Possible Moves
moves = {
    0: [1, 3],
    1: [0, 2, 4],
    2: [1, 5],
    3: [0, 4, 6],
    4: [1, 3, 5, 7],
    5: [2, 4, 8],
    6: [3, 7],
    7: [4, 6, 8],
    8: [5, 7]
}

# Generate Next States
def next_state(state):
    zero = state.index(0)
    result = []

    for move in moves[zero]:
        new = state[:]
        new[zero], new[move] = new[move], new[zero]
        result.append(new)

    return result

# Breadth-First Search
def bfs(start):
    queue = deque()
    queue.append((start, []))
    visited = []

    while queue:
        state, path = queue.popleft()

        if state == goal:
            return path + [state]

        if state not in visited:
            visited.append(state)

            for s in next_state(state):
                if s not in visited:
                    queue.append((s, path + [state]))

    return None

# Initial State
start = [1, 2, 3,
         4, 5, 6,
         0, 7, 8]

# Solve Puzzle
solution = bfs(start)

# Display Solution
print("Solution Steps:\n")

if solution:
    for step in solution:
        print(step[0:3])
        print(step[3:6])
        print(step[6:9])
        print()
else:
    print("No solution found.")
```
